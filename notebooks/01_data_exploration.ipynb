```python
import pandas as pd
import numpy as np

pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)

print("Libraries loaded successfully.")
```

```python
campaigns = pd.read_csv('../data/campaigns.csv')
customers = pd.read_csv('../data/customers.csv')
events = pd.read_csv('../data/events.csv')
products = pd.read_csv('../data/products.csv')
transactions = pd.read_csv('../data/transactions.csv')
```

These findings will guide data cleaning and attribution analysis in subsequent notebooks.

```python
print("Campaign duplicates:", campaigns.duplicated().sum())
print("Customer duplicates:", customers.duplicated().sum())
print("Event duplicates:", events.duplicated().sum())
print("Product duplicates:", products.duplicated().sum())
print("Transaction duplicates:", transactions.duplicated().sum())
```
