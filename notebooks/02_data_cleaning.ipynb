## Stage 2: Data Cleaning and Preparation

### Overview
This stage focuses on preparing the raw marketing attribution datasets for analysis. The goal is to ensure data quality, consistency, and analytical readiness before performing customer journey and attribution modeling.

### What is being evaluated in this stage?

In this stage, the following data quality dimensions are assessed and addressed:

- **Missing Data Handling**
  - Identify and treat missing values in Events and Transactions datasets
  - Determine whether to impute, retain, or flag missing records

- **Data Type Standardization**
  - Convert date fields from string format to datetime format
  - Ensure numeric and categorical fields are correctly typed

- **Data Consistency**
  - Align product_id formats across datasets
  - Standardize boolean and categorical fields

- **Data Integrity Checks**
  - Validate transactional completeness
  - Ensure event and customer relationships remain intact

- **Feature Readiness**
  - Prepare datasets for customer journey and attribution analysis
  - Create clean, analysis-ready datasets

### Outcome of this Stage
At the end of this stage, all datasets will be cleaned, standardized, and ready for:
- Customer journey mapping
- Marketing attribution modeling
- KPI and ROI analysis

```python
# Handle missing values in 'device_type' column, replacing 40,300 rows out of N2m rows NaN with 'Unknown'
events["device_type"] = events["device_type"].fillna("Unknown")
```

```python
# Handle missing values in 'product_id' column, replacing 20,000 rows out of N2m rows NaN with 'No_Product'
events["product_id"] = events["product_id"].fillna("No_Product")
```

```python
# Convert 'timestamp' column to datetime format for accurate time-based analysis
events["timestamp"] = pd.to_datetime(events["timestamp"])
```

```python
# Create a new column 'transaction_status' in the transactions dataset to indicate whether the transaction is valid or invalid based on the presence of 'gross_revenue'
transactions["transaction_status"] = transactions["gross_revenue"].apply(
    lambda x: "valid" if pd.notnull(x) else "invalid"
)
```

### Investigate transaction quality
The transactions dataset shows that missing product_id and missing gross_revenue is 10, 499 rows which is correspondingly the same.
Thus, this definition:
Valid transactions:
product_id present
gross_revenue present

Invalid/incomplete transactions:
missing product_id OR revenue
